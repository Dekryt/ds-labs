Задание по варианту 26: найти сумму различных простых делителей числа

Объявление функции решения задания по варианту:

```python
# Функция решения задания по варианту:
def sum_prime_divisors(n):
    mass = set()
    result = 0
    for i in range(2, n + 1):
      while (n % i == 0): # если деление без остатка
        mass.add(i)
        n /= i;
    for i in mass:
        result += i
    return result
```

Пример №1 работы функции:

```python
# Изменяемая переменная для тестирования:
test_value = 42

print(f"Сумма простых делителей числа {test_value} равна {sum_prime_divisors(test_value)}")
```

Пример №2 работы функции:

```python
# Изменяемая переменная для тестирования:
test_value = 84

print(f"Сумма простых делителей числа {test_value} равна {sum_prime_divisors(test_value)}")
```
